```python
# استدعاء مكتبة NLTK الخاصة بمعالجة اللغة الطبيعية
import nltk

# استدعاء مكتبة string للتعامل مع الأحرف والرموز الخاصة
import string

# تحميل مجموعة بيانات "punkt" من NLTK، وهي تستخدم لتقسيم النصوص إلى جمل وكلمات (Tokenization)
nltk.download('punkt')

# تحميل قائمة "stopwords" من NLTK، وهي كلمات شائعة مثل "the" و "and" غالبًا لا تُحسب في التحليل
nltk.download('stopwords')

# تحميل بيانات "punkt_tab" من NLTK، وهي نسخة محسنة من tokenizer للتعامل مع علامات التبويب في النصوص
nltk.download('punkt_tab')
```

```python
text = "hello, i'm moustafa software engineer ,app developer ,and Ai learner"
sentences = nltk.sent_tokenize(text,language="english")
print(sentences)
```

```python
tokens = nltk.word_tokenize(text,language = "english")
print(tokens)
```

```python
clean_tokens = [i for i in tokens if i not in string.punctuation ]
print(clean_tokens)
```
